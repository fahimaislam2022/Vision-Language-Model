#16 Generating a Histogram
An image histogram is a discrete statistical representation of the frequency distribution
of intensity values. It maps each possible intensity level (0 to 255) to the number of pixels
in the image that possess that specific intensity. Analyzing the histogram allows us to
mathematically determine if an image is overexposed, underexposed, or lacks contrast.

```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
histogram = cv2.calcHist([image_gray], [0], None, [256], [0, 256])
print("Histogram Vector Shape:", histogram.shape)
print("Most frequent intensity level index:", histogram.argmax())
plt.plot(histogram, color='black')
plt.title("16. Grayscale Pixel Histogram")
plt.xlim([0, 256])
plt.show()
```

#17 Spatial Padding (Adding Borders)
Padding involves artificially expanding the spatial dimensions of an image matrix by
adding rows and columns to the borders. These new pixels are typically filled with zeros
(zero-padding), constants, or reflections of the image edge. Padding is mathematically
vital for convolution operations to ensure the filter kernel can operate on edge pixels
without causing dimensionality reduction.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
padded_image = cv2.copyMakeBorder(image_rgb, 20, 20, 20, 20, cv2.BORDER_CONSTANT, value=[255, 0, 0])
print('Original Shape:', image_rgb.shape)
print('Padded Shape:', padded_image.shape)
plt.imshow(padded_image)
plt.title("17. Padded Image (Red Border)")
plt.axis('on')
plt.show()
```

#18 Region Masking (Modifying Specific Pixel Regions)
Masking is a localized operation where a specific region of the image is targeted for
manipulation while the rest remains untouched. It relies on a secondary binary matrix
(the mask) of the same spatial dimensions. By multiplying the image matrix with the
mask, areas corresponding to zeros in the mask are blacked out, isolating the specific
spatial region of interest.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
masked_image = np.copy(image_rgb)
masked_image[50:150, 50:150] = [0, 0, 0]
print("Masked Region Center Pixel:", masked_image[100, 100])
print("Unmasked Region Pixel:", masked_image[10, 10])
plt.imshow(masked_image)
plt.title("18. Masked Image (Black Region)")
plt.axis('on')
plt.show()
```

#19. Matrix Channel Swapping (RGB to BGR via Array Manipulation)

while other legacy systems and OpenCV natively use BGR (Blue, Green, Red). Channel
swapping is the tensor operation of reordering the depth indices to ensure color channels

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
swapped_image = image_rgb[:, :, ::-1]
print('Original Pixel (0, 0) RGB:', image_rgb[0, 0])
print('Swapped pixel (0, 0) BGR:', swapped_image[0, 0])
plt.imshow(swapped_image)
plt.title('19. Swapped Channels (BGR Displayed as RGB)')
plt.axis('on')
plt.show()
```

#20. Logical Operations (Bitwise AND with a Mask)
Bitwise operations apply Boolean logic (AND, OR, XOR, NOT) at the binary bit level of
each pixel’s integer value. A bitwise AND operation between an image and a binary mask
acts as a perfect mathematical sieve: f(x, y) AND 1 = f(x, y), while f(x, y) AND 0 = 0.
This is the standard computational method for extracting non-rectangular shapes from
an image grid.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
mask = np.zeros(image_rgb.shape[:2], dtype='uint8')
cv2.circle(mask, (image_rgb.shape[1]//2, image_rgb.shape[0]//2), 100, 255, -1)
bitwise_image = cv2.bitwise_and(image_rgb, image_rgb, mask=mask)
print("Mask Shape:", mask.shape)
print("Output Shape:", bitwise_image.shape)
print("Center Pixel Value:", bitwise_image[image_rgb.shape[0]//2, image_rgb.shape[1]//2])
plt.imshow(bitwise_image)
plt.title('20.Bitwise AND (Circular Mask)')
plt.axis('on')
plt.show()
```

#21. 1. Custom Matrix Convolution (Low-Pass Filtering)

Convolution is the cornerstone of spatial image processing. It involves sliding a small

mathematical matrix (called a kernel or filter) over the larger image matrix. At each po-
sition, a dot product is calculated between the kernel and the underlying pixels. A kernel

filled with equal positive fractions acts as a low-pass filter, averaging local neighborhoods
to mathematically blur the image and suppress high-frequency noise.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
kernel_size = 15
convolution_kernel = np.ones((kernel_size, kernel_size), np.float32) / (kernel_size * kernel_size)
convolved_image = cv2.filter2D(image_rgb, -1, convolution_kernel)
print('Original matrix Shape:', image_rgb.shape)
print('Convolution Kernel Shape:', convolution_kernel.shape)
print('Kernel Matrix Elements:\n', convolution_kernel)
print('Center Pixel Before Convolution:', image_rgb[image_rgb.shape[0]//2, image_rgb.shape[1]//2])
print('Center Pixel After Convolution:', convolved_image[convolved_image.shape[0]//2, convolved_image.shape[1]//2])
plt.imshow(convolved_image)
plt.title('21.Custom Convolution (Blur)')
plt.axis('on')
plt.show()
```

# 22. Spatial Gradient Magnitude (Sobel Edge Detection)
Edges in an image represent areas of rapid intensity change. Mathematically, this is the
first derivative of the spatial function. The Sobel operator uses two convolution kernels
to approximate the discrete spatial derivatives in the horizontal (Gx) and vertical (Gy)
directions. The overall edge strength is calculated as the gradient magnitude:  
(G2x + G2y)^1/2

```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
sobel_x = cv2.Sobel(image_gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(image_gray, cv2.CV_64F, 0, 1, ksize=3)
gradient_magnitude = np.sqrt(np.square(sobel_x) + np.square(sobel_y))
gradient_magnitude_normalized = np.uint8(255 * gradient_magnitude / np.max(gradient_magnitude))
print('Sobel X Matrix Data Type:', sobel_x.dtype)
print('Sobel Y Matrix Data Type:', sobel_y.dtype)
print('Maximum Gradient Magnitude Before Normalization:', np.max(gradient_magnitude))
print('Minimum Gradient Magnitude Before Normalization:', np.min(gradient_magnitude))
print('Normalized Output Shape:', gradient_magnitude_normalized.shape)
plt.imshow(gradient_magnitude_normalized, cmap='gray')
plt.title('22.Sobel Gradient Magnitude')
plt.axis('on')
plt.show()
```

#23. Non-Linear Intensity Transformation (Gamma Correction)

Digital displays and human vision do not process luminance linearly. Gamma correction
is a non-linear power-law transformation defined as Vout = A · V
γ
in. If γ < 1, dark regions
are brightened (encoding); if γ > 1, the image is darkened (decoding). This mathematical
curve maps linear camera sensor data to visually pleasing human perceptual scales.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
gamma_value = 2.2
inverse_gamma = 1.0 / gamma_value
lookup_table = np.array([((i / 255.0) ** inverse_gamma) * 255 for i in np.arange(0, 256)]).astype('uint8')
gamma_corrected_image = cv2.LUT(image_rgb, lookup_table)
print('Applied Gamma Value:', gamma_value)
print('Lookup Table Length:', len(lookup_table))
print('Original Intensity at 100:', 100)
print('Mapped Intensity for 100:', lookup_table[100])
print('Original Mean Intensity:', np.mean(image_rgb))
print('Corrected Mean Intensity:', np.mean(gamma_corrected_image))
plt.imshow(gamma_corrected_image)
plt.title('23.Gamma Correction (Gamma = 2.2)')
plt.axis('on')
plt.show()
```

#24. Adaptive Contrast via Histogram Equalization in YCrCb
Standard Histogram Equalization applies a global transformation, which can wash out details in locally bright or dark regions. Contrast Limited Adaptive Histogram Equal-ization (CLAHE) solves this by mathematically dividing the image matrix into small,overlapping tiles. It equalizes the histogram within each local tile and interpolates the
boundaries, ensuring local details are enhanced without global blowout.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_ycrcb = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2YCrCb)
y_channel, cr_channel, cb_channel = cv2.split(image_ycrcb)
equalized_y_channel = cv2.equalizeHist(y_channel)
merged_ycrcb = cv2.merge((equalized_y_channel, cr_channel, cb_channel))
equalized_rgb = cv2.cvtColor(merged_ycrcb, cv2.COLOR_YCrCb2RGB)
print("Original Y Channel Variance:", np.var(y_channel))
print("Equalized Y Channel Variance:", np.var(equalized_y_channel))
print("Original Y Channel Min/Max:", np.min(y_channel), np.max(y_channel))
print("Equalized Y Channel Min/Max:", np.min(equalized_y_channel), np.max(equalized_y_channel))
plt.imshow(equalized_rgb)
plt.title("24. Y- Channel Histogram Equalization")
plt.axis('on')
plt.show()
```

#25.2D Affine Transformation (Rotation Matrix)

An affine transformation is a linear mapping method that preserves points, straight lines,
and parallel lines. Mathematically, it is represented by a 2×3 transformation matrix that
multiplies the original (x, y) coordinate vectors to compute new spatial locations. This
singular mathematical framework allows for simultaneous translation, rotation, scaling,
and shearing of the image grid.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
height, width = image_rgb.shape[:2]
rotation_center = (width / 2, height / 2)
rotation_angle = 45
scale_factor = 1.0
affine_matrix = cv2.getRotationMatrix2D(rotation_center, rotation_angle, scale_factor)
rotated_image = cv2.warpAffine(image_rgb, affine_matrix, (width, height))
print('Rotation Center( X, Y):', rotation_center)
print('Rotation Angle (Degrees):', rotation_angle)
print('Computed 2x3 Affine Matrix:\n', affine_matrix)
print("Output Canvas Dimensionss:", width, "x", height)
print("Interpolated Data Type:", rotated_image.dtype)
plt.imshow(rotated_image)
plt.title("25. 45-Degree Affine Rotation")
plt.axis('on')
plt.show()
```

#26. Perspective Geometric Transformation (Homography)
Unlike affine transformations, perspective transformations (homography) do not preserve
parallelism. They simulate the distortion of viewing a 2D plane from an arbitrary 3D camera angle. Mathematically, it requires a 3 × 3 transformation matrix applied to homogeneous coordinates, introducing a division by depth to create the illusion of fore-shortening and true 3D perspective warp.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
height, width = image_rgb.shape[:2]
source_points = np.float32([[0, 0], [width-50, 0], [0, height], [width, height]])
destination_points = np.float32([[50, 50], [width-50, 0], [0, height-50], [width-50, height-50]])
perspective_matrix = cv2.getPerspectiveTransform(source_points, destination_points)
warped_image = cv2.warpPerspective(image_rgb, perspective_matrix, (width, height))
print('Source Coordinate Matrix:\n', source_points)
print('Destination Coordinate Matrix:\n', destination_points)
print('Computed 3x3 Homography Matrix:\n', perspective_matrix)
print('Determinant of Homography Matrix:', np.linalg.det(perspective_matrix))
plt.imshow(warped_image)
plt.title('26.Perspective Warping')
plt.axis('on')
plt.show()
```

#27 Statistical Additive Gaussian Noise
Gaussian noise is statistical noise having a probability density function equal to that of
the normal distribution. It mathematically simulates the random electronic fluctuations
found in digital camera sensors. Adding a matrix of random values sampled from a
Gaussian distribution (N(μ, σ2

)) to the image tensor introduces this high-frequency noise,

which is heavily used for stress-testing AI vision models.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
row, col, ch = image_rgb.shape
mean_val = 0
standard_deviation = 25
gaussian_noise = np.random.normal(mean_val, standard_deviation, (row, col, ch))
gaussian_noise = gaussian_noise.reshape(row, col, ch)
noisy_image = image_rgb.astype(np.float32) + gaussian_noise
noisy_image_clipped = np.clip(noisy_image, 0, 255).astype(np.uint8)
print('Noise Distributuin Mean:', mean_val)
print('Noise Distribution StdDev:', standard_deviation)
print('Noise matrix Shape:', gaussian_noise.shape)
print('Maximum Noise Value Inserted:', np.max(gaussian_noise))
print('Minimum Noise Value Inserted:', np.min(gaussian_noise))
print('Final Matrix Data Type Post-Clipping:', noisy_image_clipped.dtype)
plt.imshow(noisy_image_clipped)
plt.title('27.Additive Gaussian Noise')
plt.axis('on')
plt.show()
```

#28. Impulse Noise (Salt & Pepper) and Median Filtering

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
corrupted_image = np.copy(image_rgb)
probability = 0.5
random_matrix = np.random.rand(corrupted_image.shape[0], corrupted_image.shape[1])
corrupted_image[random_matrix < (probability / 2)] = [0, 0, 0]
corrupted_image[random_matrix > (1 - probability / 2)] = [255, 255, 255]
filtered_image = cv2.medianBlur(corrupted_image, 5)
print('Total Pixels Evaluted:', corrupted_image.shape[0] * corrupted_image.shape[1])
print('Impulse Probability :', probability)
print('Median Filter Kernel Size:', 5)
print('Mean Absolute Error (Original Vs Corrupted):', np.mean(np.abs(image_rgb.astype(float) - corrupted_image.astype(float))))
print('Mean Absolute Error (Original vs Filtered):', np.mean(np.abs(image_rgb.astype(float) - filtered_image.astype(float))))
plt.imshow(filtered_image)
plt.title('28 Median Filtered Image')
plt.axis('on')
plt.show()
```

#29. HSV Color Space Conversion and Saturation Mapping

The RGB model is additive but unintuitive for describing human color perception. The
HSV (Hue, Saturation, Value) color model separates the color information (Hue and
Saturation) from the luminance (Value). Mathematically converting the RGB tensor
to an HSV tensor makes it substantially easier to isolate a specific color (like "yellow")
regardless of environmental lighting conditions.


```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)
hue_channel = image_hsv[:, :, 0]
saturation_channel = image_hsv[:, :, 1]
value_channel = image_hsv[:, :, 2]
print('HSV Matrix Shape:', image_hsv.shape)
print('Hue Min/Max (0-179 in OpenCV):', np.min(hue_channel), np.max(hue_channel))
print('Saturation Min/Max (0-255):', np.min(saturation_channel), np.max(saturation_channel))
print('Value Min/Max (0-255):', np.min(value_channel), np.max(value_channel))
print('Average Image Saturation:', np.mean(saturation_channel))
plt.imshow(saturation_channel, cmap='plasma')
plt.title('29. Saturation Heatmap (HSV Space)')
plt.axis('on')
plt.show()
```

#31. Morphological Operations (Erosion and Dilation)
Mathematical morphology relies on set theory to analyze and process geometric structures
in an image (usually binary). The two foundational operations are Erosion (which strips
away boundary pixels, reducing object size) and Dilation (which adds boundary pixels,
7
expanding objects). Combining these allows for the mathematical removal of internal
holes or external speckle noise.


```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
_, binary_mask = cv2.threshold(image_gray, 127, 255, cv2.THRESH_BINARY_INV)
structuring_element = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (11, 11))
eroded_matrix = cv2.erode(binary_mask, structuring_element, iterations=2)
dilated_matrix = cv2.dilate(eroded_matrix, structuring_element, iterations=2)
print('Structuring Element Shape:', structuring_element.shape)
print('Structuring Element Array:\n', structuring_element)
print('Erosion Iteration Applied:', 2)
print('Dilation Iterations Applied:', 2)
print('Active Pixels in Original Mask:', np.count_nonzero(binary_mask))
print('Active Pixels Post-Morphology:', np.count_nonzero(dilated_matrix))
plt.imshow(dilated_matrix, cmap='gray')
plt.title('31. Morophological Opening (Ellipse Kernel)')
plt.axis('on')
plt.show()
```

Singular Value Decomposition (SVD) is an advanced linear algebra theorem stating that
any matrix can be factored into three distinct matrices: A = UΣV
T
. For images, the
diagonal matrix Σ holds the "singular values" representing the variance. By mathematically truncating this matrix to keep only the largest values, we can perfectly reconstruct
the dominant structural components of the image while discarding massive amounts of
memory.

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
matrix_float = np.float64(image_gray)
U, S, Vt = np.linalg.svd(matrix_float, full_matrices=False)
k_singular_values = 30
compressed_matrix = np.dot(U[:, :k_singular_values], np.dot(np.diag(S[:k_singular_values]), Vt[:k_singular_values, :]))
compressed_image = np.clip(compressed_matrix, 0, 255).astype(np.uint8)
print('Original Matrix Shape:', matrix_float.shape)
print('U Matrix Shape:', U.shape)
print('Singular Value Array (S) Shape:', S.shape)
print('V-Transpose Matrix Shape:', Vt.shape)
print('Retained Singular Value (k):', k_singular_values)
print('Highest Singular Value:', S[0])
plt.imshow(compressed_image, cmap='gray')
plt.title('34.SVD Compression (k=30)')
plt.axis('off')
plt.show()
```

#35. Frequency Domain Analysis (Fast Fourier Transform)
The Fast Fourier Transform (FFT) translates the spatial image matrix into the frequency
domain. Instead of representing pixels in X, Y space, the image is mathematically decomposed into a sum of 2D sine and cosine waves of varying frequencies and amplitudes.
This domain allows for pure mathematical isolation of textures (high frequencies) and
smooth gradients (low frequencies).


```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
fourier_transform = np.fft.fft2(image_gray)
shifted_fourier = np.fft.fftshift(fourier_transform)
magnitude_spectrum = 20 * np.log(np.abs(shifted_fourier) + 1)
magnitude_spectrum = np.asarray(magnitude_spectrum, dtype=np.uint8)
print("Complex FFT Matrix Shape:", fourier_transform.shape)
print("Data Type of FFT output:", fourier_transform.dtype)
print("Zero-Frequency Component (DC) location before shift:", (0, 0))
print('Zero-Frequency Component (DC) location after shift:', (shifted_fourier.shape[0]//2, shifted_fourier.shape[1]//2))
print('Max Magnitude Spectrum Value:', np.max(magnitude_spectrum))
plt.imshow(magnitude_spectrum, cmap='gray')
plt.title('35.FFT Magnitude Spectrum')
plt.axis('on')
plt.show()
```

#39. Harris Corner Detection and Tensor Marking
Corners are areas in an image where gradients change rapidly in multiple directions
simultaneously. The Harris Corner mathematical model calculates a local covariance
matrix of derivatives for a sliding window. By analyzing the eigenvalues of this matrix,
the algorithm assigns a mathematical score to each pixel, accurately identifying robust
geometric corners regardless of rotation or lighting

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
gray_float = np.float32(image_gray)
harris_response = cv2.cornerHarris(gray_float, 2, 3, 0.04)
harris_response_dilated = cv2.dilate(harris_response, None)
marked_corners = image_rgb.copy()
threshold_value = 0.01 * harris_response_dilated.max()
marked_corners[harris_response_dilated > threshold_value] = [255, 0, 0]
print('Input Tensor Data Type:', gray_float.dtype)
print('Harris Response Matrix Shape:', harris_response.shape)
print('Max Harris Response Value:', harris_response.max())
print('Applied Reponse Threshold:', threshold_value)
print('Estimated Corner Density:', np.count_nonzero(harris_response_dilated > threshold_value))
plt.imshow(marked_corners)
plt.title('39.Harris Corner Detection')
plt.axis('on')
plt.show()
```

#42. Cumulative Histogram Matching (Exact CDF Mapping)

```python
image_bgr = cv2.imread('cat.png')
source_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
target_gray = cv2.equalizeHist(source_gray)
src_hist, _ = np.histogram(source_gray.flatten(), 256, [0, 256])
tgt_hist, _ = np.histogram(target_gray.flatten(), 256, [0, 256])
src_cdf = src_hist.cumsum()
tgt_cdf = tgt_hist.cumsum()
src_cdf_normalized = src_cdf * float(tgt_cdf.max()) / src_cdf.max()
lookup_table = np.zeros(256, dtype=np.uint8)
for src_pixel_val in range(256):
  differences = np.abs(tgt_cdf - src_cdf_normalized[src_pixel_val])
  closest_match = np.argmin(differences)
  lookup_table[src_pixel_val] = closest_match
matched_image = cv2.LUT(source_gray, lookup_table)
print('Source CDF Max Value:', src_cdf.max())
print('Target CDF Max Value:', tgt_cdf.max())
print('Lookup Table Length:', len(lookup_table))
print('Mapping for Intensity 50:', lookup_table[50])
print('Mapping for Intensity 200:', lookup_table[200])
print('Means Intensity of Matched Image:', np.mean(matched_image))
plt.imshow(matched_image, cmap='gray')
plt.title('42. Histogram Matched Image')
plt.axis('on')
plt.show()
```

#43. Gabor Filter Bank for Texture Extraction

```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
gabor_kernels = []
ksize = 31
for theta in np.arange(0, np.pi, np.pi / 4):
  kernel = cv2.getGaborKernel((ksize, ksize), 4.0, theta, 10.0, 0.5, 0, ktype=cv2.CV_32F)
  kernel /= 1.5 * kernel.sum()
  gabor_kernels.append(kernel)
filtered_images = []
for kernel in gabor_kernels:
  filtered = cv2.filter2D(image_gray, cv2.CV_8UC3, kernel)
  filtered_images.append(filtered)
maximum_response = np.max(np.array(filtered_images), axis=0)
print('Number of Gabor Kernels Generated:', len(gabor_kernels))
print('Kernel Matrix Shape:', gabor_kernels[0].shape)
print('Maximum Filter Response Value:', np.max(maximum_response))
print('Minimum Filter Response Value:', np.min(maximum_response))
print('Output Response Tensor Shape:', maximum_response)
plt.imshow(maximum_response, cmap='hot')
plt.title('43.Gabor Texture Maximum Response')
plt.axis('on')
plt.show()
```
