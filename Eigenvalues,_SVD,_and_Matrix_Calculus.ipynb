Computing Eigenvalues

```python
import numpy as np
```

```python
A=np.array([[4,2],[2,3]])
print (A)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(A)
print (eigenvalues)
print (eigenvectors)
```

```python
for i in range (len(eigenvalues)):
  v=eigenvectors[:,i]
  lhs=A@v
  rhs=eigenvalues[i]*v
  error=np.linalg.norm(lhs-rhs)
  print (
      f"\n Eigenvalues {eigenvalues[i]:4f}"
      f"||Av-v||={error:2e}"
  )
  print (
      f"\n Trace:{np.trace(A):4f}|"
      f"Sum of eigenvalues:{np.sum(eigenvalues):4f}"
  )
  print (
      f"\n Determiant:{np.linalg.det(A):4f}|"
      f"Product of eigenvalues:{np.prod(eigenvalues):4f}"
  )
```

Eigendecomposition

```python
V=eigenvectors
Lambda=np.diag(eigenvalues)
A_reconstructed = V @ Lambda @ np.linalg.inv(V)
print("General reconstruction error :", np.linalg.norm(A - A_reconstructed))

A_reconstructed_sym = V @ Lambda @ V.T
print("Symmetric reconstruction error :", np.linalg.norm(A - A_reconstructed_sym))
```

Power Iteration

```python
def power_iteration(A, num_iters=100):
    n = A.shape[0]
    v = np.random.randn(n)
    v = v / np.linalg.norm(v)

    for _ in range(num_iters):
        v = A @ v
        v = v / np.linalg.norm(v)

    eigenvalue = (v.T @ A @ v) / (v.T @ v)
    return eigenvalue, v

eigval, eigvec = power_iteration(A)
print(f"Power iteration = {eigval:.4f}")
dominant_idx = np.argmax(np.abs(eigenvalues))
print(f"Numpy result = {eigenvalues[dominant_idx]:.4f}")
```

Full and Thin SVD

```python
import numpy as np

np.random.seed(42)
A = np.random.randn(5, 3)
print("Matrix A (5x3):")
print(A)
```

```python
U, S, Vt = np.linalg.svd(A, full_matrices=True)

print(f"\nU shape: {U.shape} | S shape: {S.shape} | Vt shape: {Vt.shape}")
print(f"\nSingular values: {S}")

Sigma = np.zeros_like(A, dtype=float)
Sigma[:len(S), :len(S)] = np.diag(S)
A_reconstructed = U @ Sigma @ Vt
print(f"\nReconstruction error: {np.linalg.norm(A - A_reconstructed):.2e}")


U_thin, S_thin, Vt_thin = np.linalg.svd(A, full_matrices=False)
print(f"\nThin U shape: {U_thin.shape}")
A_reconstructed_thin = U_thin @ np.diag(S_thin) @ Vt_thin
print(f"Thin reconstruction error: {np.linalg.norm(A - A_reconstructed_thin):.2e}")
```
