```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open('cat.png')
A = np.array(img)
print(f"Image Tensor Shape: {A.shape}")

alpha = 1.2
beta = 30
A_prime = np.clip(alpha * A + beta, 0, 255).astype(np.uint8)
plt.imshow(A_prime)
plt.title("Linearly Transformed Image")
plt.show()

w = np.array([0.2989, 0.5870, 0.1140])
A_gray = np.dot(A[..., :3], w)
plt.imshow(A_gray, cmap='gray')
plt.title("Grayscale via Dot Product")
plt.show()

X = A_gray
U, S, Vt = np.linalg.svd(X, full_matrices=False)
k = 50
A_k = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))
plt.imshow(A_k, cmap='gray')
plt.title(f"Reconstructed Image with k={k}")
plt.show()
```

# 3. Matrix Transposition (Diagonal Reflection)

# 4. Horizontal Flip (Matrix Reversal)

```python
img = Image.open('cat.png')
A = np.array(img)

A_flipped = A[:, ::-1, :]

plt.imshow(A_flipped)
plt.title("Horizontally Flipped")
plt.show()
```

# 5. Matrix Slicing (Image Cropping)

```python
img = Image.open('cat.png')
A = np.array(img)

h, w, _ = A.shape
A_cropped = A[h//4:3*h//4, w//4:3*w//4, :]

plt.imshow(A_cropped)
plt.title("Cropped Center")
plt.show()
```

# 6. Sepia Filter (Matrix Multiplication)

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

sepia_matrix = np.array([
    [0.393, 0.769, 0.189],
    [0.349, 0.686, 0.168],
    [0.272, 0.534, 0.131]
])

A_sepia = np.dot(A, sepia_matrix.T)
A_sepia = np.clip(A_sepia, 0, 255).astype(np.uint8)

plt.imshow(A_sepia)
plt.title("Sepia Filter")
plt.show()
```

# 7. Binary Thresholding (Logical Matrix Operations)

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

w = np.array([0.2989, 0.5870, 0.1140])
A_gray = np.dot(A, w)

threshold = 128
A_binary = (A_gray > threshold) * 255
A_binary = A_binary.astype(np.uint8)

plt.imshow(A_binary, cmap='gray')
plt.title("Binary Thresholding")
plt.show()
```

# 8. Adding Gaussian Noise (Matrix Addition)

```python
img = Image.open('cat.png')
A = np.array(img)

noise = np.random.normal(0, 25, A.shape)
A_noisy = A.astype(np.float32) + noise
A_noisy = np.clip(A_noisy, 0, 255).astype(np.uint8)

plt.imshow(A_noisy)
plt.title("Gaussian Noise")
plt.show()
```

# 9. Image Blending (Linear Combination)

```python
img = Image.open('cat.png')
A = np.array(img)
A_flipped = A[:, ::-1, :]

alpha = 0.5
A_blended = (alpha * A + (1 - alpha) * A_flipped).astype(np.uint8)

plt.imshow(A_blended)
plt.title("Blended Image")
plt.show()
```

# 11. Color Channel Masking (Red Channel Only)

# 12. Matrix Normalization (Min-Max Scaling)

```python
img = Image.open('cat.png')
A = np.array(img).astype(np.float32)

A_min = np.min(A)
A_max = np.max(A)
A_norm = (A - A_min) / (A_max - A_min)

plt.imshow(A_norm)
plt.title("Normalized Image")
plt.show()
```

# 13. Posterization (Color Quantization via Matrix Division)

```python
img = Image.open('cat.png')
A = np.array(img)

levels = 4
factor = 256 // levels
A_posterized = (A // factor) * factor

plt.imshow(A_posterized)
plt.title("Posterized Image")
plt.show()
```

# 14. Simple Edge Detection (Matrix Shift and Subtract)

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

w = np.array([0.2989, 0.5870, 0.1140])
A_gray = np.dot(A, w)

A_shift_x = np.roll(A_gray, 1, axis=1)
A_shift_y = np.roll(A_gray, 1, axis=0)

edges = np.abs(A_gray - A_shift_x) + np.abs(A_gray - A_shift_y)
edges = np.clip(edges, 0, 255).astype(np.uint8)

plt.imshow(edges, cmap='gray')
plt.title("Shift-and-Subtract Edges")
plt.show()
```

# 15. Vignette Effect (Hadamard Product with a Radial Mask)

```python
img = Image.open('cat.png')
A = np.array(img)

h, w = A.shape[:2]
Y, X = np.ogrid[:h, :w]
center_y, center_x = h / 2, w / 2

dist_from_center = np.sqrt((X - center_x)**2 + (Y - center_y)**2)
max_dist = np.sqrt(center_x**2 + center_y**2)

mask = 1 - (dist_from_center / max_dist)
mask = np.clip(mask, 0, 1)[..., np.newaxis]

A_vignette = (A * mask).astype(np.uint8)

plt.imshow(A_vignette)
plt.title("Vignette Effect")
plt.show()
```

# 16. Image Flattening and Reshaping (Vectorization)

```python
img = Image.open('cat.png')
A = np.array(img)

h, w, c = A.shape
A_flat = A.reshape(-1, c)
A_reshaped = A_flat.reshape(h, w, c)

plt.imshow(A_reshaped)
plt.title("Flattened and Reshaped")
plt.show()
```

# 17. Gamma Correction (Power Law Transformation)

```python
img = Image.open('cat.png')
A = np.array(img).astype(np.float32) / 255.0

gamma = 2.2
A_gamma = np.power(A, 1.0 / gamma)
A_gamma = (A_gamma * 255).astype(np.uint8)

plt.imshow(A_gamma)
plt.title(f"Gamma Correction (g={gamma})")
plt.show()
```

# 18. Fast Box Blur (Matrix Shifting and Addition)

```python
img = Image.open('cat.png')
A = np.array(img).astype(np.float32)

A_blur = (
    A[1:-1, 1:-1] +
    A[:-2, 1:-1] + A[2:, 1:-1] +
    A[1:-1, :-2] + A[1:-1, 2:]
) / 5.0

A_blur = A_blur.astype(np.uint8)

plt.imshow(A_blur)
plt.title("Simple Fast Blur (Matrix Shift)")
plt.show()
```

# 19. Color Isolation (Euclidean Distance in Color Space)

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3].astype(np.float32)

target_color = np.array([255, 0, 0])
distances = np.linalg.norm(A - target_color, axis=2)

mask = (distances < 150)[..., np.newaxis]
A_isolated = (A * mask).astype(np.uint8)

plt.imshow(A_isolated)
plt.title("Color Isolation (Red Proximity)")
plt.show()
```

# 20. RGB Singular Value Decomposition (SVD per Channel)

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

k = 30
A_compressed = np.zeros_like(A)

for i in range(3):
    U, S, Vt = np.linalg.svd(A[..., i], full_matrices=False)
    A_compressed[..., i] = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))

A_compressed = np.clip(A_compressed, 0, 255).astype(np.uint8)

plt.imshow(A_compressed)
plt.title(f"RGB SVD Compression (k={k})")
plt.show()
```

# 21. Multi-k RGB Compression Comparison

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

k_values = [10, 50, 150]
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for idx, k in enumerate(k_values):
    A_k = np.zeros_like(A)
    for i in range(3):
        U, S, Vt = np.linalg.svd(A[..., i], full_matrices=False)
        A_k[..., i] = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))

    A_k = np.clip(A_k, 0, 255).astype(np.uint8)
    axes[idx].imshow(A_k)
    axes[idx].set_title(f"k={k}")
    axes[idx].axis('off')

plt.show()
```

# 22. Rank-1 RGB Approximation

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

A_rank1 = np.zeros_like(A)

for i in range(3):
    U, S, Vt = np.linalg.svd(A[..., i], full_matrices=False)
    A_rank1[..., i] = S[0] * np.outer(U[:, 0], Vt[0, :])

A_rank1 = np.clip(A_rank1, 0, 255).astype(np.uint8)

plt.imshow(A_rank1)
plt.title("Rank-1 RGB Approximation")
plt.show()
```

# 23. Plotting RGB Singular Value Decay

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

colors = ['r', 'g', 'b']

for i in range(3):
    _, S, _ = np.linalg.svd(A[..., i], full_matrices=False)
    plt.plot(S, color=colors[i], label=f'Channel {colors[i].upper()}')

plt.yscale('log')
plt.title("Singular Values Decay (RGB)")
plt.legend()
plt.show()
```

# 24. RGB SVD Mean Squared Error (MSE) Calculation

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

k = 20
A_comp = np.zeros_like(A, dtype=np.float32)

for i in range(3):
    U, S, Vt = np.linalg.svd(A[..., i], full_matrices=False)
    A_comp[..., i] = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))

A_comp = np.clip(A_comp, 0, 255)

mse = np.mean((A.astype(np.float32) - A_comp) ** 2)

plt.imshow(A_comp.astype(np.uint8))
plt.title(f"RGB SVD (k={k}) | MSE: {mse:.2f}")
plt.show()
```

# 25. Channel-wise Reconstructed Outputs

```python
img = Image.open('cat.png')
A = np.array(img)[..., :3]

k = 25
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
titles = ['Red Channel', 'Green Channel', 'Blue Channel']

for i in range(3):
    U, S, Vt = np.linalg.svd(A[..., i], full_matrices=False)
    channel_comp = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))

    disp_img = np.zeros_like(A)
    disp_img[..., i] = np.clip(channel_comp, 0, 255)

    axes[i].imshow(disp_img)
    axes[i].set_title(titles[i])
    axes[i].axis('off')

plt.show()
```
